```python
import pandas as pd
import numpy as np

# MultiIndex
index = [
    ('A', 2010),
    ('A', 2020),
    ('A', 2025),
    ('B', 2010),
    ('B', 2020),
    ('B', 2025)
]

population = [123, 456, 789, 101, 121, 131]

i = pd.MultiIndex.from_tuples(index)

print(i)

# Create Series using MultiIndex
pop = pd.Series(population, index=i)

print("\nOriginal Series:")
print(pop)

# Convert inner index (year) into columns
pop_df = pop.unstack()

print("\nAfter unstack():")
print(pop_df)

# Convert back to Series
print("\nAfter stack():")
print(pop_df.stack())

# MultiIndex DataFrame
df = pd.DataFrame(
    np.random.rand(4, 2),
    index=pd.MultiIndex.from_tuples([
        ('A', 2010),
        ('A', 2020),
        ('B', 2010),
        ('B', 2020)
    ])
)

print("\nDataFrame:")
print(df)

# Dictionary with tuple keys
data = {
    ('A', 2010): 123,
    ('A', 2020): 456,
    ('A', 2025): 789,
    ('B', 2010): 101,
    ('B', 2020): 121
}

print("\nSeries from dictionary:")
print(pd.Series(data))
```

```python
i1 = pd.MultiIndex.from_arrays(
    [
        ['A', 'A', 'A', 'B', 'B', 'B'],
        [2010, 2020, 2025, 2010, 2020, 2025]
    ],
    names=["Country", "Year"]
)

pop1 = pd.Series(population, index=i1)

print(pop1)
```

```python
i2 = pd.MultiIndex.from_product(
    [['A', 'B'], [2010, 2020, 2025]],
    names=["Country", "Year"]
)

pop2 = pd.Series(population, index=i2)

print(pop2)
```

```python
i3=pd.MultiIndex(levels=[['A','B'],[2010,2020,2025]],codes=[[0,0,0,1,1,1],[0,1,2,0,1,2]],names=["Country","Year"])
pop3=pd.Series(population,index=i3)
print(pop3)

```
