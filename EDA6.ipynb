```python
import numpy as np
import pandas as pd
data=pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
data.dropna()
```

```python
data=pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
print(data.isnull())
print(data.notnull())
```

```python
data3=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data3)
print(data3.dropna())
print(data3.dropna(axis='columns'))
print(data3.dropna(axis='columns',how='any'))
print(data3.dropna(axis='rows',thresh=3))
```

```python
data4=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data4)
print(data4.fillna(0))
print(data4.ffill())
print(data4.bfill())
```
