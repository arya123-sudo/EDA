```python
#1
import numpy as np
x=np.random.randint(1,100,10)
print("orginal array:",x)
sorted=np.sort(x)
print("using sort:",sorted)
x.sort()
print(x)
#2
i=np.argsort(x)
print(i)
result=x[i]
print("x[i]:",result)
print("np.sort:",sorted)
print("Are they same:",np.array_equal(result,sorted))
```

```python
#3
x=np.random.randint(1,100,(4,6))
print(x)
row_sorted=np.sort(x,axis=1)
cloumn_sorted=np.sort(x,axis=0)
print("row wise:",row_sorted)
print("col wise:",cloumn_sorted)

```

```python
x=np.random.randint(1,100,15)
print("orginal array: ",x)
result=np.partition(x,4)
print("partition method:",result)
print("smallest 4 integer:",result[:4])
res=np.sort(x)
print("using sort function:",res)
print("smallest 4 integer:",res[:4])
```

```python
#5
import numpy as np

student_names = np.array([
    "Rahul",
    "Priya",
    "Arjun",
    "Sneha",
    "Kiran"
])

scores = np.array([75, 92, 68, 85, 78])
i = np.argsort(scores)
ranked_names = student_names[i]
print("Scores:", scores)
print("Students ranked from lowest to highest:")

for name, score in zip(ranked_names, scores[i]):
    print(name, ":", score)
```
